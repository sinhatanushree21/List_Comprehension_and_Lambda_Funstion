**List Comprehension**: List comprehension is a concise, one-line syntax in Python used to create a new list by iterating over an existing iterable (like a list, tuple, or range). It essentially replaces traditional for loops and append() setups, making your code shorter and often faster


Basic Syntax:
new_list = [expression for item in iterable if condition]

**Using a traditional for loop:**

```python
square = []
for i in range(1,7):
  square.append (i*i)
print(square)
```

**Using a list comprehension:**

**Why Use It?**

**Readability:** It condenses multiple lines of loop logic into a clean, single line.

**Performance:** It runs faster than manual .append() loops because it is optimized inside Python's C-level engine

```python
square = [i*i for i in range(0,7)]
print(square)
```

**Adding Filters (if Condition)**

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = [num for num in numbers if num % 2 == 0]
evens
```

```python
labels = ['even' if x%2 == 0 else 'odd' for x in range(0,7)]
labels
```

A **lambda** function in Python is a small, anonymous function that can have any number of arguments, but can only have one single expression. It is called "anonymous" because it is a function defined without a name (unlike the standard def keyword).

**Key Characteristics**

**No return keyword:** You cannot use the return statement. The result of the expression is returned automatically.

**Single expression only:** You cannot write multiple lines of logic or loops inside a lambda.

**Inline usage:** They are designed to be written inline right where you need them, without needing to assign them to a variable name first.

**Basic Syntax**

lambda arguments: expression

**Using standard def syntax:**

```python
def add_ten(x):
    return x + 10

print(add_ten(5))
```

**Using lambda syntax:**

```python
add_ten = lambda x : x + 10
print(add_ten(5))
```

```python
square = lambda x : x*x
print(square(5))
```

```python
add = lambda x : x+x
print(add(5))
```

```python
add = lambda a,b : a+b
print(add(5,4))
```

```python
# A lambda function that multiplies two numbers
multiply = lambda a, b: a * b

print(multiply(4, 5))
```

**Sorting Lists of Tuples or Dictionaries (The key parameter)**

```python
students = [("Alice", 25), ("Bob", 20), ("Charlie", 23)]

students = sorted(students, key=lambda x: x[1])
students
```
